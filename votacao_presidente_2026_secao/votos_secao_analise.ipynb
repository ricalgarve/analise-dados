# Votação por Seção Eleitoral - 2026

Dados abertos do Tribunal Superior Eleitoral (TSE).
Fonte dataset: https://dadosabertos.tse.jus.br/dataset/resultados-2026

```python
import pandas as pd

# Carregando o dataset do TSE (usa separador ';' e codificacao 'latin1')
df = pd.read_csv('votacao_secao_2026_BR.csv', sep=';', encoding='latin1')

# Remove o limite máximo de colunas exibidas na tela
pd.set_option('display.max_columns', None)

# Exibindo as primeiras linhas
df.head()
```

Escolas de Pederneiras

```python
## escolas de pederneiras buscar 

df_pederneiras = df[df['NM_MUNICIPIO'] == 'PEDERNEIRAS'].copy()

# Listar todas as escolas com endereço e seções
escolas_pederneiras = (
    df_pederneiras.groupby(['NR_LOCAL_VOTACAO', 'NM_LOCAL_VOTACAO', 'DS_LOCAL_VOTACAO_ENDERECO'])['NR_SECAO']
    .nunique()
    .reset_index(name='TOTAL_SECOES')
    .sort_values(by='TOTAL_SECOES', ascending=False)
)
escolas_pederneiras
```

Levantamento das seções da escola João Chamas 



```python
# Filtrar dados da Escola E.E. Comendador João Chammas (código 1082)
df_joao_chammas = df_pederneiras[df_pederneiras['NR_LOCAL_VOTACAO'] == 1082]

# Lista ordenada com as seções eleitorais da escola
secoes_joao_chamas_pederneiras = sorted(df_joao_chammas['NR_SECAO'].unique())

print(f'Total de seções: {len(secoes_joao_chamas_pederneiras)}')
print('Seções:', secoes_joao_chamas_pederneiras)

# Exibir os dados das seções com candidatos e votos
df_result_votos_secao = df_joao_chammas[['NM_LOCAL_VOTACAO', 'NR_SECAO', 'NR_VOTAVEL', 'NM_VOTAVEL', 'QT_VOTOS']]

df_result_votos_secao.head(10)
```

```python

# 1. Configura o Pandas para exibir TODAS as linhas na tela sem truncar (...)
pd.set_option('display.max_rows', None)

# 2. Agrupamento completo por Seção e Nome do Presidente (somando os votos)
# (Você pode usar df_pederneiras para toda a cidade ou df_joao_chammas para a escola)
df_todos_votos_secao = (
    df_pederneiras.groupby(['NM_LOCAL_VOTACAO', 'NR_SECAO', 'NM_VOTAVEL'])['QT_VOTOS']
    .sum()
    .reset_index()
    .sort_values(by=['NR_SECAO', 'QT_VOTOS'], ascending=[True, False])
)

# 3. Exibir todos os dados
#df_todos_votos_secao = df_todos_votos_secao[df_todos_votos_secao['NR_SECAO'] == 82]

df_todos_votos_secao.head()
```

Relação de votos gerais por município Pederneiras

```python
# 1. Agrupamento dos votos por candidato em todo o município
votos_gerais_todas_secoes = (
    df_todos_votos_secao.groupby('NM_VOTAVEL')['QT_VOTOS']
    .sum()
    .reset_index()
    .sort_values(by='QT_VOTOS', ascending=False)
)

# 2. Calcular total de votos da cidade e percentuais
total_votos_cidade = votos_gerais_todas_secoes['QT_VOTOS'].sum()
votos_gerais_todas_secoes['PERCENTUAL'] = (
    (votos_gerais_todas_secoes['QT_VOTOS'] / total_votos_cidade) * 100
).round(2).astype(str) + '%'

# 3. Linha totalizadora no final
linha_total = pd.DataFrame([{
    'NM_VOTAVEL': 'TOTAL DE VOTOS NA CIDADE',
    'QT_VOTOS': total_votos_cidade,
    'PERCENTUAL': '100.00%'
}])

# 4. Tabela final com totalizador
votos_gerais_com_total = pd.concat([votos_gerais_todas_secoes, linha_total], ignore_index=True)

votos_gerais_com_total
```

## Comparativo: Top Escolas de Pederneiras - Flávio Bolsonaro vs. Lula

Avaliação do total de votos e percentuais obtidos por cada candidato em todas as escolas do município.

```python
# 1. Total de votos de cada escola
total_escola = df_pederneiras.groupby('NM_LOCAL_VOTACAO')['QT_VOTOS'].sum()

# 2. Tabela com candidatos nas colunas
votos_escolas = df_pederneiras.groupby(['NM_LOCAL_VOTACAO', 'NM_VOTAVEL'])['QT_VOTOS'].sum().unstack(fill_value=0)
votos_escolas['TOTAL_ESCOLA'] = total_escola

# 3. Ranking das Escolas - Flávio Bolsonaro
ranking_flavio = votos_escolas[['FLAVIO NANTES BOLSONARO', 'TOTAL_ESCOLA']].copy()
ranking_flavio['PERCENTUAL'] = ((ranking_flavio['FLAVIO NANTES BOLSONARO'] / ranking_flavio['TOTAL_ESCOLA']) * 100).round(2).astype(str) + '%'
ranking_flavio = ranking_flavio.sort_values(by='FLAVIO NANTES BOLSONARO', ascending=False).reset_index()

# 4. Ranking das Escolas - Lula
ranking_lula = votos_escolas[['LUIZ INÁCIO LULA DA SILVA', 'TOTAL_ESCOLA']].copy()
ranking_lula['PERCENTUAL'] = ((ranking_lula['LUIZ INÁCIO LULA DA SILVA'] / ranking_lula['TOTAL_ESCOLA']) * 100).round(2).astype(str) + '%'
ranking_lula = ranking_lula.sort_values(by='LUIZ INÁCIO LULA DA SILVA', ascending=False).reset_index()

print('=== TOP ESCOLAS - FLÁVIO BOLSONARO ===')
display(ranking_flavio)

print('\n=== TOP ESCOLAS - LULA ===')
display(ranking_lula)
```
