## 9. Pandas: Estrutura Series

As *Series* são arrays unidimensionais rotulados, capazes de armazenar diferentes tipos de dados.

## 12. Seleção e Exclusão de Variáveis (Colunas)

````python
# Selecionando uma única coluna (Series)
classe = titanic['classe']

# Selecionando múltiplas colunas (novo DataFrame)
selecao_var = titanic[['sobreviveu', 'classe', 'sexo']]
selecao_var.head()
````

````python
# Excluindo variáveis usando o método drop(columns=[...])
selecao_var_sem_classe = selecao_var.drop(columns=['classe'])
print("Sem classe:")
display(selecao_var_sem_classe.head(2))

# Excluindo múltiplas colunas do DataFrame original selecao_var
selecao_var_reduzida = selecao_var.drop(columns=['classe', 'sobreviveu'])
print("Apenas sexo:")
display(selecao_var_reduzida.head(2))
````

## 13. Filtrando Observações (Linhas)

Principais operadores de comparação e lógica:
- `==` igual | `!=` diferente
- `>` maior | `>=` maior ou igual
- `<` menor | `<=` menor ou igual
- `&` operador E (AND) | `|` operador OU (OR)

````python
# 1. Filtro: Passageiros que sobreviveram (duas opções equivalentes)
sobreviveu = titanic[titanic['sobreviveu'] == 'sim']
# sobreviveu = titanic[titanic['sobreviveu'] != 'nao']
sobreviveu.info()
````

````python
# 2. Filtro: Adultos do sexo masculino
masc_adultos = titanic[(titanic['sexo'] == 'masculino') & (titanic['idade'] >= 18)]
print(f"Total de homens adultos: {len(masc_adultos)}")

# 3. Filtro: Mulheres OU crianças (idade < 18)
mulheres_criancas = titanic[(titanic['sexo'] == 'feminino') | (titanic['idade'] < 18)]
print(f"Total de mulheres ou crianças: {len(mulheres_criancas)}")

# 4. Filtro: Pessoas sozinhas (sem cônjuges, irmãos, pais ou filhos)
sozinhos = titanic[(titanic['irmaos_conjuges'] == 0) & (titanic['pais_filhos'] == 0)]
print(f"Total de passageiros sozinhos: {len(sozinhos)}")

# 5. Filtro: Passageiros que pagaram tarifa superior a $100 e embarcaram em Cherbourg
tarifas_embarque = titanic[(titanic['valor_tarifa'] > 100) & (titanic['embarque'] == 'Cherbourg')]
print(f"Tarifa > 100 e embarque Cherbourg: {len(tarifas_embarque)}")
````

## 14. Limpeza de Dados (Tratamento de Dados Faltantes)

- Se usarmos diretamente `dropna()`, muitas linhas são perdidas devido à coluna `nivel_cabine` (composta majoritariamente por valores ausentes).
- Portanto, remove-se primeiro a coluna `nivel_cabine` antes de descartar os valores nulos restantes.

````python
# Excluindo a variável 'nivel_cabine' e removendo observações faltantes restantes
titanic_limpo = titanic.drop(columns=['nivel_cabine']).dropna()
titanic_limpo.info()
````

## 15. Estatísticas Descritivas: Frequências

Contagem absoluta e relativa (`normalize=True`) da sobrevivência:

````python
# Frequência absoluta de sobreviventes
print("--- Frequência Absoluta ---")
print(titanic['sobreviveu'].value_counts())

# Frequência relativa (proporção)
print("--- Frequência Relativa Geral ---")
print(titanic['sobreviveu'].value_counts(normalize=True))

# Sobrevivência entre os que viajavam sozinhos
print("--- Sobrevivência (Sozinhos) ---")
print(sozinhos['sobreviveu'].value_counts(normalize=True))

# Sobrevivência entre homens adultos
print("--- Sobrevivência (Homens Adultos) ---")
print(masc_adultos['sobreviveu'].value_counts(normalize=True))
````

## 16. Estatísticas Descritivas: Variável Quantitativa (Idade e Tarifa)

Cálculo pontual das principais medidas de tendência central e de dispersão:

````python
print("Contagem:", titanic['idade'].count())
print("Média aritmética:", titanic['idade'].mean())
print("Mediana:", titanic['idade'].median())
print("Máximo:", titanic['idade'].max())
print("Mínimo:", titanic['idade'].min())
print("1º Quartil (Q1 - 25%):", titanic['idade'].quantile(q=0.25))
print("3º Quartil (Q3 - 75%):", titanic['idade'].quantile(q=0.75))
print("Variância:", titanic['idade'].var())
print("Desvio padrão:", titanic['idade'].std())
````

````python
# Resumo estatístico completo com .describe()
print("Resumo da Idade:")
display(titanic['idade'].describe())

print("Resumo da Idade e Tarifa:")
display(titanic[['idade', 'valor_tarifa']].describe())

print("Resumo de todas as variáveis numéricas do dataset:")
display(titanic.describe())
````

## 17. Estatísticas Descritivas Agrupadas (`groupby`)

Análise de métricas por categorias qualitativas:

````python
# 1. Valor médio da tarifa por classe
print("--- Tarifa Média por Classe ---")
display(titanic[['classe', 'valor_tarifa']].groupby(by=['classe']).mean())

# 2. Percentual de sobreviventes por sexo
print("--- Percentual de Sobreviventes por Sexo ---")
display(titanic[['sobreviveu', 'sexo']].groupby('sexo')['sobreviveu'].value_counts(normalize=True))

# 3. Percentual de sobreviventes por classe
print("--- Percentual de Sobreviventes por Classe ---")
display(titanic[['sobreviveu', 'classe']].groupby('classe')['sobreviveu'].value_counts(normalize=True))
````

## 18. Visualização de Dados (Data Visualization)

Utilizaremos as bibliotecas `matplotlib.pyplot` e `seaborn` para exploração visual.

````python
import matplotlib.pyplot as plt
import seaborn as sns

# Recarregando o banco de dados conforme o fluxo do script
titanic = pd.read_excel('titanic.xlsx')
````

### 18.1. Gráfico de Contagem (Countplot) - Variável Qualitativa

````python
# Gráfico de contagem básico
plt.figure(figsize=(10, 6))
sns.countplot(data=titanic, x='sobreviveu')
plt.title('Contagem Básica de Sobreviventes', fontsize=16)
plt.show()
````

````python
# Gráfico formatado com rótulos e cores personalizadas
plt.figure(figsize=(10, 6))
ax = sns.countplot(data=titanic, x='sobreviveu', color='purple')

# Adicionando rótulos com a contagem exata no topo de cada barra
for container in ax.containers:
    ax.bar_label(container, fontsize=12)

plt.title('Titanic: Contagem de Sobreviventes', fontsize=18)
plt.xlabel('Sobreviveu?', fontsize=14)
plt.ylabel('Contagem', fontsize=14)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.show()
````

````python
# Detalhando sobrevivência por sexo (parâmetro hue)
plt.figure(figsize=(10, 6))
ax = sns.countplot(data=titanic, x='sobreviveu', hue='sexo', palette='viridis', legend=True)

for container in ax.containers:
    ax.bar_label(container, fontsize=12)

plt.title('Titanic: Sobrevivência por Sexo', fontsize=18)
plt.xlabel('Sobreviveu?', fontsize=14)
plt.ylabel('Contagem', fontsize=14)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.show()
````

### 18.2. Gráfico de Barras com Agregação (Barplot)

Comparando o valor médio da tarifa por cidade de embarque:

````python
# Agrupando para calcular a média da tarifa por embarque
dados = titanic[['embarque', 'valor_tarifa']].groupby(by=['embarque']).mean()

plt.figure(figsize=(10, 6))
ax1 = sns.barplot(data=dados, x=dados.index, y='valor_tarifa', hue=dados.index, palette='bright')

for container in ax1.containers:
    ax1.bar_label(container, fmt='%.2f', padding=3, fontsize=12)

plt.title("Tarifa Média por Local de Embarque", fontsize=18)
plt.xlabel('Cidade de Embarque', fontsize=14)
plt.ylabel('Valor Médio ($)', fontsize=14)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.show()
````

### 18.3. Histograma - Variável Métrica (Idade)

Visualização da distribuição de frequência da idade dos passageiros:

````python
plt.figure(figsize=(10, 6))
sns.histplot(data=titanic, x='idade', bins=range(0, 85, 5), kde=True, color='lightgreen')
plt.title('Distribuição da Idade dos Passageiros (com KDE)', fontsize=18)
plt.xlabel('Idade', fontsize=14)
plt.ylabel('Frequência', fontsize=14)
plt.xticks(ticks=np.arange(0, 85, 5), fontsize=12)
plt.yticks(fontsize=12)
plt.show()
````

### 18.4. Gráfico de Dispersão (Scatterplot & Regplot)

Relação entre Idade e Valor da Tarifa (para tarifas abaixo de $100):

````python
# Dispersão simples e reta de regressão
plt.figure(figsize=(10, 6))
sns.regplot(data=titanic[titanic['valor_tarifa'] < 100], x='idade', y='valor_tarifa',
            scatter_kws={'alpha': 0.6}, line_kws={'color': 'red'})
plt.title('Relação Idade x Tarifa (< $100) com Linha de Tendência', fontsize=16)
plt.xlabel('Idade', fontsize=13)
plt.ylabel('Valor da Tarifa ($)', fontsize=13)
plt.show()
````

````python
# Dispersão categorizada pela classe do bilhete
plt.figure(figsize=(11, 6))
sns.scatterplot(
    data=titanic[titanic['valor_tarifa'] < 100],
    x='idade',
    y='valor_tarifa',
    hue='classe',
    hue_order=['primeira', 'segunda', 'terceira'],
    palette='Set1',
    alpha=0.8
)
plt.title('Titanic: Idade vs Tarifa (< $100) por Classe', fontsize=18)
plt.xlabel('Idade', fontsize=14)
plt.ylabel('Valor da Tarifa ($)', fontsize=14)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.show()
````

### 18.5. Gráfico de Linhas (Lineplot)

Tarifa média por classe do bilhete:

````python
dados_linha = titanic[['classe', 'valor_tarifa']].groupby(by=['classe']).mean()

plt.figure(figsize=(10, 6))
sns.lineplot(data=dados_linha, x=dados_linha.index, y='valor_tarifa', marker='o', linewidth=3, color='purple', markersize=8)
plt.title('Tarifa Média por Classe', fontsize=18)
plt.xlabel('Classes', fontsize=14)
plt.ylabel('Tarifa Média ($)', fontsize=14)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
````

### 18.6. Boxplot com Medidas de Posição

````python
plt.figure(figsize=(10, 7))
sns.boxplot(y=titanic['idade'], color='orange', width=0.3)

# Calculando medidas estatísticas
minimo = titanic['idade'].min()
q1 = titanic['idade'].quantile(q=0.25)
q2 = titanic['idade'].median()
q3 = titanic['idade'].quantile(q=0.75)
maximo = titanic['idade'].max()

# Inserindo anotações com os valores calculados
plt.text(0.2, minimo, f'Mín = {minimo:.1f}', ha='left', va='center', fontweight='bold', fontsize=11)
plt.text(0.2, q1, f'Q1 = {q1:.1f}', ha='left', va='center', fontweight='bold', fontsize=11)
plt.text(0.2, q2, f'Mediana (Q2) = {q2:.1f}', ha='left', va='center', fontweight='bold', color='blue', fontsize=11)
plt.text(0.2, q3, f'Q3 = {q3:.1f}', ha='left', va='center', fontweight='bold', fontsize=11)
plt.text(0.2, maximo, f'Máx = {maximo:.1f}', ha='left', va='center', fontweight='bold', fontsize=11)

plt.title('Boxplot da Idade com Medidas de Posição', fontsize=18)
plt.ylabel('Idade', fontsize=14)
plt.show()
````
