# Análise de Mortalidade e Expectativa de Vida no Brasil

## O que é este projeto
- Um estudo prático sobre como e com que idade os brasileiros estão morrendo, usando dados reais do SUS e do IBGE.
- A meta é transformar registros brutos de óbito em curvas de mortalidade e estimativas de expectativa de vida por região e faixa etária.

## Quais dados vamos usar
- **SIM (Sistema de Informações sobre Mortalidade / DataSUS):** dados de cada óbito registrado (idade, sexo, causa da morte pela CID-10, município e escolaridade).
- **IBGE (População residente):** censo e estimativas anuais de quantas pessoas vivem em cada faixa etária (necessário para calcular a proporção real de óbitos por habitante).
- **PySUS:** biblioteca Python para puxar esses arquivos direto dos servidores do governo sem precisar baixar arquivos na mão pelo navegador.

## Período analisado e anos recentes
- **2015 a 2023:** período com dados 100% consolidados pelo Ministério da Saúde. Cobre o padrão pré-pandemia, o impacto da COVID-19 (2020-2021) e a recuperação recente.
- **2024 e 2025:** são bases preliminares (ainda em processo de digitação e qualificação de laudos nos municípios). Servem para acompanhar tendências recentes, mas não para fechar tábuas de vida definitivas.
- **Estratégia de início:** testar primeiro com um estado (ex: SP ou RJ) e um único ano recente para entender as colunas e validar o código antes de puxar o Brasil inteiro.

## Passo a passo planejado
1. Instalar o PySUS e checar a conexão com o DataSUS.
2. Baixar os microdados de um ano e estado de teste.
3. Tratar a idade (o DataSUS usa códigos especiais para horas, dias, meses e anos de vida).
4. Levantar a distribuição de mortes por idade, sexo e principais causas.
5. Montar a tábua de vida simplificada e comparar o resultado com as projeções do IBGE.
